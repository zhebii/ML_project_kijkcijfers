# Verslag en reflectie

## Doel

Het doel van dit project is om het aantal kijkers voor televisieprogramma’s te voorspellen op basis van historische kijkcijfers en weersomstandigheden. Door patronen te ontdekken in hoe weer en andere factoren invloed hebben op kijkgedrag, willen we een model ontwikkelen dat toekomstige kijkcijfers zo accuraat mogelijk kan inschatten.<br><br>
Het project combineert data-analyse en machine learning om inzichten te verkrijgen in het kijkgedrag van televisiekijkers, met het uiteindelijke doel om deze kijkcijfers automatisch te kunnen voorspellen.

## Ophalen en verwerken data

Voor dit project werden twee externe databronnen gebruikt: een open-meteo API om historische weergegevens op te halen, en een andere API om televisiekijkcijfers te verkrijgen.<br><br>
De opgehaalde datasets zijn vervolgens gefilterd zodat enkel geldige en relevante waarden werden behouden bij het wegschrijven naar CSV-bestanden. Bij de verwerking van de kijktijddata werden de tijdsnotaties opgeschoond: uren die boven 24 uitkwamen werden gecorrigeerd. Verder werd de startTime-kolom opgesplitst in drie afzonderlijke kolommen: FullDate, hour en minute, zodat er makkelijker mee gewerkt kon worden tijdens de analyse.<br><br>
Om de leesbaarheid te verbeteren, zijn de kolomnamen hernoemd naar duidelijkere benamingen. Nadat beide datasets grondig opgeschoond waren, zijn ze samengevoegd tot één grote tabel. Deze gecombineerde dataset vormde de basis voor verdere analyse en modelbouw.

## Model training en finetuning

Tijdens de modelbouwfase werden extra kenmerken (features) toegevoegd aan de dataset, zoals isFeestdag, Weekdag, isWeekend en Seizoen. Deze extra variabelen boden meer context over de omstandigheden waarin programma’s bekeken werden, wat belangrijk is voor zowel analyse als voorspelling.<br><br>
Vervolgens werd de dataset voorbereid voor machine learning. Kolommen met een lage cardinaliteit (zoals dag van de week) werden verwerkt met OneHotEncoding, terwijl kolommen met een hoge cardinaliteit (zoals programmanamen of zenders) via Target Encoding werden omgezet. De resulterende samengestelde tabel van deze encoders diende als input voor de modeltraining.<br><br>
Voor de modelselectie werden meerdere modellen getest in een vergelijkende testrun. Hieruit werden de drie best presterende modellen geselecteerd en verder gefinetuned. Voor de hyperparameteroptimalisatie werden zowel Randomized Search als Optuna gebruikt om de prestaties verder te verbeteren.<br><br>
Het uiteindelijke beste model (LGBMRegressor met optuna) werd opgeslagen als een .pkl-bestand (pickle-formaat), zodat het later eenvoudig gebruikt kon worden in de voorspelling pipeline.

## Resultaat

Het uiteindelijke geoptimaliseerde model behaalde een Mean Absolute Error (MAE) van ongeveer 50.000, wat betekent dat het model gemiddeld 50.000 kijkers naast de werkelijke waarde zat. Ondanks deze afwijking slaagde het model erin om voorspellingen te maken met een nauwkeurigheid van ongeveer 85%.

## Conclusie

Het doel van dit project was om te onderzoeken of het mogelijk is om het aantal kijkers van televisieprogramma’s te voorspellen op basis van historische kijkcijfers en weersinformatie. Door systematisch data te verzamelen, op te schonen en te verrijken met extra kenmerken zoals feestdagen, seizoenen en weekdagen, kon er een solide dataset worden opgebouwd voor modeltraining.<br><br>
Na het testen van meerdere machine learning-modellen en het finetunen van hyperparameters via technieken zoals Randomized Search en Optuna, bleek LightGBM het beste te presteren met een MAE van ongeveer 50.000 en een voorspelnauwkeurigheid rond de 85%. Hoewel dit geen perfect resultaat is, toont het wel aan dat er een significante voorspelbaarheid zit in de patronen van kijkgedrag.
